Q4. what products are the most often sold together

```python
all_data.head()
```

```python
df= all_data[all_data['Order ID'].duplicated(keep = False)]
df.head()
```

```python
df['Grouped'] = df.groupby('Order ID')['Product'].transform(lambda x: ','.join(x))
df2 = df[['Order ID', 'Grouped']].drop_duplicates()

df2.head()
```

```python
from itertools import combinations
from collections import Counter

count = Counter()

for row in df2['Grouped']:
    row_list = row.split(',')
    count.update(Counter(combinations(row_list, 2)))

for key,value in count.most_common(10):
    print(key, value)
```

Q5.  What product sold the most ? why do you think it sold the most ?

```python
all_data.head()
```

```python
product_group=all_data.groupby('Product')['Quantity Ordered'].sum()
```

```python
plt.bar(product_group.index,product_group.values)
plt.ylabel('Quantity Ordered')
plt.xlabel('Product')
plt.xticks(product_group.index,rotation='vertical',size = 8)
plt.show()
```

```python
price=all_data.groupby('Product')['Price Each'].mean()
fig, ax1 = plt.subplots()

ax2 = ax1.twinx()
ax1.bar(product_group.index,product_group.values, color='g')
ax2.plot(product_group.index, price.values, color='b')

ax1.set_xlabel('Product Name')
ax1.set_ylabel('Quantity Ordered', color='g')
ax2.set_ylabel('Price ($)', color='b')
ax1.set_xticklabels(product_group.index, rotation='vertical', size=8)

fig.show()
```
